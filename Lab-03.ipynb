## Task 1: Security System Scan

**Scenario**
- A security system has 9 components (A to I).
- Each component is either Safe or Vulnerable.

**Goal**
- Scan every component, identify vulnerabilities, and patch them.

**Steps**
1. **Initial system check**
   - Initialize components with random vulnerabilities.
   - Display which components are Safe vs Vulnerable.
2. **System scan**
   - Scan each component.
   - Log a warning for Vulnerable components.
   - Log success for Safe components.
3. **Patching**
   - Patch all Vulnerable components and mark them Safe.
   - Display which components were patched.
4. **Final system check**
   - Display the final state (all components Safe).

```python
import random

class Environment:

    def __init__(self):
        global components
        components = []
        for i in range(1,10):
            val = random.randint(0,1)
            if val == 0:
                components.append(f"{i}: Secure")
            elif val == 1:
                components.append(f"{i}: Vulnerable")

    def getState(self):
        print(components)


class Agent(Environment):

    def __init__(self):
        super().__init__()

    def scanComponents(self):
        global to_patch
        to_patch = []
        for elem in components:
            if elem.endswith("Vulnerable"):
                to_patch.append(elem)
                print(f"Added component {elem[0]} to patch list.")
            else:
                print(f"Component {elem[0]} is secure")

        print("\nPatch List:")
        print(to_patch)            

    def patchVulns(self):
        components = []
        for i in range(1,10):
            components.append(f"{i}: Secure")
        print("\nAll vulnerabilities fixed.")
        print("\nCurrent state:")
        print(components)
        print()




# main 

print("\n" + "=" * 50)
print("🔒 Cybersecurity Auditing Agent")
print("=" * 50)
print("Starting agent...\n")
worker = Agent()
print("Initial system check...\n")
print("📋 Current component states:")
worker.getState()
print("\n" + "=" * 50)
print("Running Component scanner...")
print("=" * 50 + "\n")
worker.scanComponents()
print("\n" + "=" * 50)
print("Fixing vulnerabilities...")
print("=" * 50)
worker.patchVulns()
print("✅ System in secure state.")
print("=" * 50 + "\n")
```

## Task 2: Load Balancing in a Data Center

**Scenario**
- A data center has 5 servers.
- Each server is Underloaded, Balanced, or Overloaded.

**Goal**
- Redistribute tasks so the overall load becomes balanced.

**Requirements**
- Create 5 servers with random load states.
- Scan the system for Overloaded servers.
- Move tasks from Overloaded servers to Underloaded servers.
- Display the updated load status for each server.

```python
import random

class Server:

    def __init__(self, name):
        self.name = name
        val = random.randint(1, 3)
        if val == 1:
            self.status = "Underloaded"
        elif val == 2:
            self.status = "Balanced"
        elif val == 3:
            self.status = "Overloaded"


    def getState(self):
        return self.status


class System:

    def __init__(self):
        self.serverList = []
        for i in range(1,6):
            self.serverList.append(Server(str(i)))


    def getServerStatus(self):
        print()
        for elem in self.serverList:
            print(f"{elem.name}: {elem.status}")
        print()



class LoadBalancerAgent:

    def __init__(self, systemID):
        self.workstation = systemID

    def balanceLoad(self):
        
        for elem in self.workstation.serverList:
            if elem.status == "Overloaded":
                elem.status = "Balanced"
            elif elem.status == "Underloaded":
                elem.status = "Balanced"
        
        

        print("Successfully Balanced Load...")

    





# main


print("=" * 50)
print("Initializing System...")
print("=" * 50)
mainframe = System()

print("\n📊 Initial Server Status:")
mainframe.getServerStatus()

print("=" * 50)
print("Starting Load Balancer Agent...")
print("=" * 50)
worker = LoadBalancerAgent(mainframe)
worker.balanceLoad()

print("\n📊 Final Server Status:")
mainframe.getServerStatus()
print("=" * 50)

# Just took the print statements from AI for proper formatting
```

**Goal**
- Retry failed backups and confirm all tasks are updated.

**Requirements**
- Create a list of backup tasks with random statuses.

- Display the updated task list after retries.

```python
import random

class ServerNetwork:

    def __init__(self):
        self.backupTasks = []
        count = 1
        self.length = random.randint(3,7)
        for i in range(self.length):
            val = random.randint(0, 1)
            if val == 0: 
                self.backupTasks.append("Completed")
            elif val == 1:
                self.backupTasks.append("Failed")
            count += 1

    
    def getTasks(self):
        count = 1
        print()
        for elem in self.backupTasks:
            print(f"{count}: {elem}")
            count += 1
        print("\n-----------------------")

    def setTasks(self, tasks):
        self.backupTasks = tasks




class BackupManagementAgent:

    def __init__(self):
        pass


    def retryTasks(self, server):
        server.backupTasks = ["Completed" for i in range(server.length)]
    



# main
server = ServerNetwork()
print("-----------------------")
print("\n🔍 Initial Backup Tasks Status:")
server.getTasks()

agent = BackupManagementAgent()
print("\n⚙️  Retrying failed backups...\n")
print("All Backups Completed Successfully.\n")
agent.retryTasks(server)

print("-----------------------")
print("✅ Updated Backup Tasks Status:")
server.getTasks()
print("\n")
```

## Task 4: Utility-Based Security Agent

**Scenario**
- A security system has 9 components (A to I).
- Each component is Safe, Low Risk Vulnerable, or High Risk Vulnerable.
- Only Low Risk vulnerabilities can be patched with the basic service.

**Goal**
- Patch Low Risk vulnerabilities and report High Risk ones.

**Steps**
1. **Initial system check**
   - Initialize components with random statuses.
   - Display the initial state.
2. **System scan**
   - Log a warning for vulnerable components.
   - Log success for Safe components.
3. **Patching**
   - Patch Low Risk vulnerabilities.
   - Report High Risk vulnerabilities as requiring premium service.
4. **Final system check**
   - Display the final state (Low Risk patched, High Risk unchanged).

```python
import random

class Environment:

    def __init__(self):
        global components
        components = []
        for i in range(1,10):
            val = random.randint(0,1)
            if val == 0:
                components.append(f"{i}: Secure")
            elif val == 1:
                risk = random.randint(0,1)
                if risk == 0:
                    components.append(f"{i}: Low-Risk Vulnerable")
                elif risk == 1:
                    components.append(f"{i}: High-Risk Vulnerable")


    def getState(self):
        print("\n" + "="*30)
        print("Current Component States:")
        print("="*30 + "\n")
        for elem in components:
            print(f"  {elem}")
        print("\n" + "="*30)


class Agent(Environment):

    def __init__(self):
        super().__init__()

    def scanComponents(self):
        global to_patch
        to_patch = []
        print("\n" + "="*30)
        print("Scanning Components...")
        print("="*30 + "\n")
        for elem in components:
            if elem.endswith("Vulnerable"):
                to_patch.append(elem)
                print(f"  ⚠️  Added component {elem[0]} to patch list.")
            else:
                print(f"  ✓ Component {elem[0]} is secure")

        print("\nPatch List:")
        print("="*30)
        for i in to_patch:
            print(f"  {i}")
        print("\n" + "="*30)

    def patchVulns(self):
        fixed = []
        count = 1
        for elem in components:
            if "Secure" in elem:
                fixed.append(f"{count}: Secure")
            elif "Low" in elem:
                fixed.append(f"{count}: Secure")
            elif "High" in elem:
                fixed.append(elem)
            count += 1
        print("\n" + "="*30)
        print("Patching Vulnerabilities...")
        print("="*30 + "\n")
        print("✓ Fixed all Low-Risk Vulnerabilities.")
        print("\n⚠️  Please subscribe to premium service to patch High-Risk Vulnerabilities.")
        print("\n" + "="*30)
        


# main

print("Initializing Systems and Agents...")
system = Environment()
agent = Agent()

system.getState()
agent.scanComponents()
agent.patchVulns()
system.getState()

# Used AI for print statements and formatting only
```

## Task 5: Hospital Delivery Robot

**Scenario**
- A robot delivers medicine and assists staff in a hospital.
- The robot navigates corridors, rooms, nurse stations, and storage areas.

**Actions**
- Move to a location.
- Pick up medicine from storage.
- Deliver medicine to a patient room.
- Scan patient ID before delivery.
- Alert staff for critical situations.

**Perceptions**
- Room numbers (delivery locations).
- Patient schedules (timing).
- Medicine type.
- Staff availability.

**Goal**
- Deliver medicines correctly based on schedule and room number.
- Always scan patient ID before delivery.
- Alert staff when needed.

```python
class Hospital:

	def __init__(self):
		self.corridors = ["Corridor-A", "Corridor-B"]
		self.rooms = {"101": {"patient_id": "P-101"}, "102": {"patient_id": "P-102"}}
		self.nurse_station_available = True
		self.storage = {"Med-A": 5, "Med-B": 5}

	def has_medicine(self, med_name):
		return self.storage.get(med_name, 0) > 0

	def take_medicine(self, med_name):
		if self.has_medicine(med_name):
			self.storage[med_name] -= 1
			return True
		return False


class DeliveryRobot:

	def __init__(self, name="Robo-1"):
		self.name = name
		self.carrying = None

	def move_to(self, location):
		print(f"\n==> {self.name} moves to {location}.")

	def pick_up(self, hospital, med_name):
		if hospital.take_medicine(med_name):
			self.carrying = med_name
			print(f"[Pickup] {self.name} picked up {med_name}.")
		else:
			print(f"[Pickup] {self.name} could not find {med_name} in storage.")

	def scan_patient_id(self, hospital, room_number, expected_id):
		actual_id = hospital.rooms[room_number]["patient_id"]
		print(f"[Scan] {self.name} scans patient ID in room {room_number}.")
		return actual_id == expected_id

	def deliver(self, room_number):
		if self.carrying is None:
			print(f"[Deliver] {self.name} has nothing to deliver.")
			return
		print(f"[Deliver] {self.name} delivers {self.carrying} to room {room_number}.")
		self.carrying = None

	def alert_staff(self, hospital, message):
		if hospital.nurse_station_available:
			print(f"[Alert] {self.name} alerts staff: {message}")
		else:
			print(f"[Alert] {self.name} waits to alert staff: {message}")

	def run_schedule(self, hospital, schedule):
		for task in schedule:
			room = task["room"]
			med = task["medicine"]
			patient_id = task["patient_id"]
			critical = task.get("critical", False)

			print("\n--- New Task ---")
			self.move_to("Medicine Storage")
			self.pick_up(hospital, med)

			self.move_to(f"Room {room}")
			if self.scan_patient_id(hospital, room, patient_id):
				self.deliver(room)
			else:
				self.alert_staff(hospital, f"ID mismatch in room {room}")

			if critical:
				self.alert_staff(hospital, f"Critical patient in room {room}")



hospital = Hospital()
robot = DeliveryRobot()

print("=" * 46)
print("Hospital Delivery Robot Simulation")
print("=" * 46)


robot.move_to("Corridor-A")
robot.move_to("Medicine Storage")
robot.pick_up(hospital, "Med-A")
robot.move_to("Room 101")
if robot.scan_patient_id(hospital, "101", "P-101"):
    robot.deliver("101")
else:
    robot.alert_staff(hospital, "ID mismatch in room 101")
print("-" * 46)

robot.move_to("Corridor-B")
robot.move_to("Medicine Storage")
robot.pick_up(hospital, "Med-B")
robot.move_to("Room 102")
if robot.scan_patient_id(hospital, "102", "P-102"):
    robot.deliver("102")
else:
    robot.alert_staff(hospital, "ID mismatch in room 102")
robot.alert_staff(hospital, "Critical patient in room 102")
print("=" * 46)


```

## Task 6: Firefighting Robot in a Building

**Scenario**
- A 3x3 building grid contains rooms that are Safe or on Fire.
- The robot starts at room 'a' and follows a fixed path to room 'j'.

**Initialization**
- Safe rooms: a, b, d, f, g, h
- Fire rooms: c, e, j
- Path: ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'j']

**Robot Movement**
- Move room to room along the path.
- If a room is on Fire, extinguish it and mark Safe.
- If a room is Safe, continue to the next room.

**Display Requirements**
- After each move, display the grid status.
- Use a clear symbol for Fire and for Safe.

**Final Output**
- After the path ends, show the final grid (all fires extinguished).

```python


class Building:

    def __init__(self):
        self.rooms = [
            [{"name": "a", "state": "Safe"}, {"name": "b", "state": "Safe"}, {"name": "c", "state": "Fire"}],
            [{"name": "d", "state": "Safe"}, {"name": "e", "state": "Fire"}, {"name": "f", "state": "Safe"}],
            [{"name": "g", "state": "Safe"}, {"name": "h", "state": "Safe"}, {"name": "j", "state": "Fire"}],
        ]


    def returnRooms(self):
        return self.rooms

    def getRoomByName(self, name):
        for floor in self.rooms:
            for room in floor:
                if room["name"] == name:
                    return room
        return None

    def display(self, current_room=None):
        for floor in self.rooms:
            row = []
            for room in floor:
                if current_room == room["name"]:
                    row.append("*")
                elif room["state"] == "Fire":
                    row.append("F")
                else:
                    row.append(".")
            print(" ".join(row))



class Firefighter:

    def __init__(self):
        self.path = ["a", "b", "c", "d", "e", "f", "g", "h", "j"]

    def extinguishFire(self, room):
        room["state"] = "Safe"
        print(f"🧯 Fire extinguished in room {room['name']}.")


    def checkRooms(self, build):
        for step, room_name in enumerate(self.path, start=1):
            room = build.getRoomByName(room_name)
            if room is None:
                print(f"[Step {step}] Room {room_name} not found.")
                continue
            
            print("\n" + "-"*50)
            print(f"\nEntering room {room_name}")
            if room["state"] == "Fire":
                print(f"🔥 Fire detected in room {room['name']}")
                self.extinguishFire(room)
            else:
                print(f"✅ Room {room['name']} is safe.")

            print("\nUpdated Status:")
            build.display(current_room=room_name)



build = Building()
robot = Firefighter()

print("\n" + "="*50)
print("\n🏢 FIREFIGHTING ROBOT - BUILDING STATUS\n")
print("="*50)

print("\n📋 Initial Status:")
print("-"*50)
build.display()

print("\n" + "="*50)
print("\n🤖 Starting Firefighting Robot\n")
print("="*50)
robot.checkRooms(build)

print("\n" + "="*50)
print("\n✨ Final Building Status (All Fires Extinguished)\n")
build.display()
print("="*50 + "\n")
```
