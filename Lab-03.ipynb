# Task 1

A company’s security system consists of nine critical components (A through I). Each
component can be either safe or vulnerable due to security flaws. A security agent is
responsible for scanning the entire system, identifying vulnerabilities, and patching
them to prevent attacks.
You are tasked with simulating a cybersecurity exercise using the given environment
and agent. Complete the following steps:
Initial System Check:
○ The system environment is initialized with random vulnerabilities.
Display the initial state of the system, showing which components are safe
and which are vulnerable.
System Scan:
○ The security agent scans each component. If a component is vulnerable,
the agent logs a warning and adds it to a list for patching. If it is secure, a
success message is logged.
Patching Vulnerabilities:
○ After the scan, the agent patches all vulnerable components, marking
them as safe. Display messages indicating which components have been
patched.
Final System Check:
○ Display the system’s final state, confirming that all vulnerabilities have
been patched.

```python
import random

class Environment:

    def __init__(self):
        global components
        components = []
        for i in range(1,10):
            val = random.randint(0,1)
            if val == 0:
                components.append(f"{i}: Secure")
            elif val == 1:
                components.append(f"{i}: Vulnerable")

    def getState(self):
        print(components)


class Agent(Environment):

    def __init__(self):
        super().__init__()

    def scanComponents(self):
        global to_patch
        to_patch = []
        for elem in components:
            if elem.endswith("Vulnerable"):
                to_patch.append(elem)
                print(f"Added component {elem[0]} to patch list.")
            else:
                print(f"Component {elem[0]} is secure")

        print("\nPatch List:")
        print(to_patch)            

    def patchVulns(self):
        components = []
        for i in range(1,10):
            components.append(f"{i}: Secure")
        print("\nAll vulnerabilities fixed.")
        print("\nCurrent state:")
        print(components)
        print()




# main 

print("\n" + "=" * 50)
print("🔒 Cybersecurity Auditing Agent")
print("=" * 50)
print("Starting agent...\n")
worker = Agent()
print("Initial system check...\n")
print("📋 Current component states:")
worker.getState()
print("\n" + "=" * 50)
print("Running Component scanner...")
print("=" * 50 + "\n")
worker.scanComponents()
print("\n" + "=" * 50)
print("Fixing vulnerabilities...")
print("=" * 50)
worker.patchVulns()
print("✅ System in secure state.")
print("=" * 50 + "\n")
```

# Task 2

A data center runs multiple servers, each hosting different services. These services are
either underloaded, balanced, or overloaded based on the system's load. The Load Balancer Agent is responsible for redistributing tasks across the servers to ensure
optimal system performance.
Task:
- Create a system with 5 servers.
- Each server has a load state of "Underloaded", "Balanced", or "Overloaded".
- The Load Balancer Agent must scan the system and move tasks from overloaded
servers to underloaded ones to balance the load.
- After balancing the load, display the updated load status of each server.

```python
import random

class Server:

    def __init__(self, name):
        self.name = name
        val = random.randint(1, 3)
        if val == 1:
            self.status = "Underloaded"
        elif val == 2:
            self.status = "Balanced"
        elif val == 3:
            self.status = "Overloaded"


    def getState(self):
        return self.status


class System:

    def __init__(self):
        self.serverList = []
        for i in range(1,6):
            self.serverList.append(Server(str(i)))


    def getServerStatus(self):
        print()
        for elem in self.serverList:
            print(f"{elem.name}: {elem.status}")
        print()



class LoadBalancerAgent:

    def __init__(self, systemID):
        self.workstation = systemID

    def balanceLoad(self):
        
        for elem in self.workstation.serverList:
            if elem.status == "Overloaded":
                elem.status = "Balanced"
            elif elem.status == "Underloaded":
                elem.status = "Balanced"
        
        

        print("Successfully Balanced Load...")

    





# main


print("=" * 50)
print("Initializing System...")
print("=" * 50)
mainframe = System()

print("\n📊 Initial Server Status:")
mainframe.getServerStatus()

print("=" * 50)
print("Starting Load Balancer Agent...")
print("=" * 50)
worker = LoadBalancerAgent(mainframe)
worker.balanceLoad()

print("\n📊 Final Server Status:")
mainframe.getServerStatus()
print("=" * 50)

# Just took the print statements from AI for proper formatting
```
