## Task 1: Agent-Based Search

Convert the following search algorithms into agent-based models:

- **Depth-Limited Search (DLS)**: implement as a **goal-based agent** that explores the graph up to a specified depth limit.
- **Uniform Cost Search (UCS)**: implement as a **utility-based agent** that finds the minimum-cost path to a goal.

```python
# DLS Agent

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H'],
    'E': [],
    'F': ['I'],
    'G': [],
    'H': [],
    'I': []
}

class Goal_Based_DLS_Agent:

    def __init__(self, graph):
        self.graph = graph

    def dls(self, start, goal, depth_limit):
        
        return self._dfs(start, goal, depth_limit, path=[])
    
    def _dfs(self, node, goal, depth_limit, path):

        path = path + [node]

        if node == goal: 
             print("--- Path Found ---")
             return path

        if depth_limit == 0:    return None

        for neighbor in self.graph.get(node, []):
            result = self._dfs(neighbor, goal, depth_limit - 1, path)
            if result is not None:
                    return result
        return None


# main

start_node = 'A'
goal_node = 'I'
depth_limit = 3

print("=== Depth-Limited Search (Goal-Based Agent) ===")
print(f"Start Node : {start_node}")
print(f"Goal Node  : {goal_node}")
print(f"Depth Limit: {depth_limit}")

ISIAgent = Goal_Based_DLS_Agent(graph)
path = ISIAgent.dls(start_node, goal_node, depth_limit=depth_limit)

if path is None:
    print("Result     : Goal not found within depth limit.")
else:
    print("Result     : Path found")
    print(f"Path       : {' -> '.join(path)}")
```

```python
# UCS Agent


graph = {
    'A': {'B': 2, 'C': 1},
    'B': {'D': 4, 'E': 3},
    'C': {'F': 1, 'G': 5},
    'D': {'H': 2},
    'E': {},
    'F': {'I': 6},
    'G': {},
    'H': {},
    'I': {}
}

class UCS_Utility_Based_Agent:

    def __init__(self, graph: dict, start: str, goals: set[str]) -> None:
        self.graph = graph
        self.start = start
        self.goals = goals

    def find_min_cost_path(self):
        frontier = [(self.start, 0)]
        visited = set()
        cost_so_far = {self.start: 0}
        came_from = {self.start: None}

        while frontier:

            frontier.sort(key=lambda x: x[1])
            current_node, current_cost = frontier.pop(0)

            if current_node in visited: continue

            visited.add(current_node)

            if current_node in self.goals:
                path = []
                node = current_node
                while node is not None:
                    path.append(node)
                    node = came_from[node]
                path.reverse()
                return current_cost, path
            
            for neighbor, edge_cost in self.graph.get(current_node, {}).items():
                new_cost = current_cost + edge_cost
                if neighbor not in cost_so_far or new_cost < cost_so_far[neighbor]:
                    cost_so_far[neighbor] = new_cost
                    came_from[neighbor] = current_node
                    frontier.append((neighbor, new_cost))
        return None, []



start_node = 'A'
goal_nodes = {'H', 'I'}

print("=" * 52)
print("Uniform Cost Search (UCS) - Utility-Based Agent")
print("=" * 52)
print(f"Start node   : {start_node}")
print(f"Goal nodes   : {', '.join(sorted(goal_nodes))}")
print("-" * 52)

RAW_agent = UCS_Utility_Based_Agent(graph, start_node, goal_nodes)
min_cost, path = RAW_agent.find_min_cost_path()

if path:
    print(f"Minimum cost : {min_cost}")
    print(f"Path         : {' -> '.join(path)}")
else:
    print("No path found to any goal node.")




```

## Task 2: Traveling Salesman Problem (TSP)

Given a set of cities and pairwise distances, find the **shortest route** that visits each city exactly once and returns to the start. The objective is to **minimize total distance**.

**Cities:** 1, 2, 3, 4

**Distances (Cost):**
- 1 - 2 = 10
- 1 - 3 = 15
- 1 - 4 = 20
- 2 - 3 = 35
- 2 - 4 = 25
- 3 - 4 = 30

**Goal:** compute the minimum-cost tour that starts and ends at city 1 (or your chosen start).

```python

from itertools import permutations


class GoalBasedTSPAgent:
	def __init__(self, distances: dict, start_city: int) -> None:
		self.distances = distances
		self.start_city = start_city

	def is_goal(self, route: tuple[int, ...]) -> bool:
		return route[0] == self.start_city and route[-1] == self.start_city

	def route_cost(self, route: tuple[int, ...]) -> int:
		total = 0
		for i in range(len(route) - 1):
			total += self.distances[route[i]][route[i + 1]]
		return total

	def find_best_route(self) -> tuple[int, tuple[int, ...]]:
		cities = [c for c in self.distances.keys() if c != self.start_city]
		best_cost = None
		best_route = ()

		for perm in permutations(cities):
			route = (self.start_city,) + perm + (self.start_city,)
			if self.is_goal(route):
				cost = self.route_cost(route)
				if best_cost is None or cost < best_cost:
					best_cost = cost
					best_route = route

		return best_cost, best_route



distances = {
    1: {2: 10, 3: 15, 4: 20},
    2: {1: 10, 3: 35, 4: 25},
    3: {1: 15, 2: 35, 4: 30},
    4: {1: 20, 2: 25, 3: 30},
}

agent = GoalBasedTSPAgent(distances, start_city=1)
best_cost, best_route = agent.find_best_route()

print("=" * 52)
print("Goal-Based Agent: Traveling Salesman Problem")
print("=" * 52)
if best_route:
    route_str = " -> ".join(str(city) for city in best_route)
    print(f"Best route : {route_str}")
    print(f"Total cost : {best_cost}")
else:
    print("No valid route found.")

```

## Task 3: Iterative Deepening DFS

Implement **Iterative Deepening DFS (IDDFS)** for both a **tree** and a **graph**, using a **goal-based agent** approach. The output should include the path to the goal if found within the depth limit.

```python
class GoalBasedIDDFSAgent:
	def __init__(self, graph: dict, start: str, goal: str) -> None:
		self.graph = graph
		self.start = start
		self.goal = goal

	def dls(self, node: str, depth: int, path: list, on_path: set) -> bool:
		if depth == 0:
			return False
		if node == self.goal:
			path.append(node)
			return True
		if node not in self.graph:
			return False

		on_path.add(node)
		for child in self.graph[node]:
			if child in on_path:
				continue
			if self.dls(child, depth - 1, path, on_path):
				path.append(node)
				return True
		on_path.remove(node)
		return False

	def iterative_deepening(self, max_depth: int) -> list:
		for depth in range(max_depth + 1):
			print(f"Depth: {depth}")
			path = []
			if self.dls(self.start, depth, path, set()):
				path.reverse()
				return path
		return []


def main():
	tree = {
		'A': ['B', 'C'],
		'B': ['D', 'E'],
		'C': ['F', 'G'],
		'D': ['H'],
		'E': [],
		'F': ['I'],
		'G': [],
		'H': [],
		'I': []
	}

	graph = {
		'A': ['B', 'C'],
		'B': ['D', 'E'],
		'C': ['F', 'G'],
		'D': ['H', 'A'],
		'E': ['B'],
		'F': ['I', 'C'],
		'G': ['C'],
		'H': ['D'],
		'I': ['F']
	}

	start_node = 'A'
	goal_node = 'I'
	max_search_depth = 5

	print("=" * 60)
	print("Iterative Deepening DFS (Goal-Based Agent)")
	print("=" * 60)
	print(f"Start node     : {start_node}")
	print(f"Goal node      : {goal_node}")
	print(f"Max depth      : {max_search_depth}")
	print("-" * 60)

	print("[Tree Search]")
	tree_agent = GoalBasedIDDFSAgent(tree, start_node, goal_node)
	tree_path = tree_agent.iterative_deepening(max_search_depth)
	if tree_path:
		print(f"Path           : {' -> '.join(tree_path)}")
	else:
		print("Path           : Not found within depth limit")

	print("-" * 60)
	print("[Graph Search]")
	graph_agent = GoalBasedIDDFSAgent(graph, start_node, goal_node)
	graph_path = graph_agent.iterative_deepening(max_search_depth)
	if graph_path:
		print(f"Path           : {' -> '.join(graph_path)}")
	else:
		print("Path           : Not found within depth limit")


if __name__ == "__main__":
	main()
```
