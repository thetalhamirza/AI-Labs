```python
import sys

count = int(input("How many values: "))
nums = []

for i in range(count):
    nums.append(int(input(f"Enter Value {i+1}: ")))

def display():
    print("")
    print("List:")
    sum = 0
    for num in nums:
        print(num)
        sum += num
    
    print("Sum:", sum)
    print("Maximum:", max(nums))
    print("Minimum:", min(nums))

display()

index = -2

while index != -1:
    index = int(input("Enter index to change (-1 to exit) : "))
    
    if index != -1:
        nums[index] = int(input("Enter the updated value: "))
        display()
    
```

# Task 4

```python
def applyDiscount(amount):
    value = amount    
    if value < 1000:
        value -= value * 0.05
    elif value < 5000:
        value -= value * 0.1
    else:
        value -= value * 0.15
    return value

amount = int(input("Enter a value: "))
print(applyDiscount(amount))
```

# Task 5

```python
def countString(text):
    chars = len(text)
    vowels = 0
    consonants = 0
    for char in text:
        if char in ('a', 'e', 'i', 'o', 'u'):
            vowels += 1
        elif char != " ":
            consonants += 1
    dictString = {"Total_Char": chars, "Vowels": vowels, "Consonants": consonants}
    return dictString

text = input("Enter a string: ")
counter = countString(text)
print("Total Characters:", counter["Total_Char"])
print("Vowels:", counter["Vowels"])
print("Consonants:", counter["Consonants"])
```

# Task 6

```python
text = input("Enter a string: ")
arr = []
output = ""

for char in text:
    if char not in arr:
        output = output + char
        arr.append(char)

print("Substring:", output)
print("Length:", len(output))
```
